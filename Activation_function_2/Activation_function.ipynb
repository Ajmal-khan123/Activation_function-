```python
import math

def sigmoid(x):
    return 1/(1 + math.exp(-x))
```

```python
sigmoid(3)
```

```python
sigmoid(100)
```

```python
sigmoid(15)
```

```python
def relu(x):
    return max(0, x)
```

```python
relu(100)
```

```python
relu(15)
```

```python
def tanh(x):
    return (math.exp(x) - math.exp(-x) / math.exp(x) - math.exp(-x) )
```

```python
tanh(100)
```

```python
tanh(15)
```

```python
def leak_relu(x):
    return max(0.1 * x, x)
```

```python
leak_relu(100)
```

```python
leak_relu(15)
```

### after visiting all the function finally we recognize that the leak_relu and relu working mostly similar.
